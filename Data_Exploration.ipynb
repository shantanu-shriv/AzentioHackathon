# Data Exploration & Cleaning
Let's start by loading our three datasets: `transactions.csv`, `accounts.csv`, and `customers.csv`. We will then inspect them for missing values, duplicates, and anomalies.

```python
import pandas as pd
import numpy as np

# Load datasets
transactions = pd.read_csv('transactions.csv')
accounts = pd.read_csv('accounts.csv')
customers = pd.read_csv('customers.csv')

print(f"Transactions shape: {transactions.shape}")
print(f"Accounts shape: {accounts.shape}")
print(f"Customers shape: {customers.shape}")
```

## 1. Inspecting Transactions Data
Let's look at the first few rows and the data types to identify any anomalies (like 'N/A' amounts or invalid timestamps).

```python
display(transactions.head())
transactions.info()
transactions.isnull().sum()
```

## 2. Inspecting Accounts Data
Checking for null values, duplicates, and corrupt credit limits.

```python
display(accounts.head())
accounts.info()
accounts.isnull().sum()
```
